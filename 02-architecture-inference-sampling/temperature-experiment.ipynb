```python
def get_api_key(identifier: str = "ANTHROPIC_API_KEY") -> str:
    """Get the Anthropic API key from Colab's userdata or a .env file."""
    try:
        from google.colab import userdata
        API_KEY = userdata.get(identifier)
    except:
        from dotenv import load_dotenv
        import os
        load_dotenv()
        API_KEY = os.getenv(identifier)
    
    return API_KEY

def print_models(client):
    """Print the list of models available in the given client."""
    for m in client.models.list():
        print(m.id, "-", m.display_name)
```

## Anthropic

```python
# Anthropic Connection test
from anthropic import Anthropic

ANTHROPIC_API_KEY = get_api_key()

anthropic_client = Anthropic(api_key=ANTHROPIC_API_KEY)

print_models(anthropic_client)
```

## Gemini

```python
# Gemini connection test
from openai import OpenAI
from google import genai
from google.genai import types


GEMINI_API_KEY = get_api_key("GEMINI_API_KEY")


gemini_client = genai.Client(api_key=GEMINI_API_KEY)

# gemini_client = OpenAI(
#     api_key=GEMINI_API_KEY,
#     base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
# )

# print_models(gemini_client)
```

```python
m = "Tell me a story about a robot learning something new."
m = "What should I search for to find the latest developments in renewable energy?"
m = "Explain the concept of quantum computing in simple terms."

response = gemini_client.models.generate_content(
    model='gemini-3.5-flash-lite',
    contents=m,
    config=types.GenerateContentConfig(
        max_output_tokens=1024,
        temperature=0.0,
    ),
)


print(response.text)
```

```python
response = gemini_client.models.generate_content(
    model='gemini-3.5-flash-lite',
    contents=m,
    config=types.GenerateContentConfig(
        max_output_tokens=1024,
        temperature=0.7,
        top_p=0.9,
    ),
)


print(response.text)
```
